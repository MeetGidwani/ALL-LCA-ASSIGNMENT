```python
# Assignment 1: Dictionary, Tuple and List of Students

# Creating a list of students
students_list = ["Rahul", "Amit", "Sneha", "Priya"]

# Creating a tuple of students
students_tuple = ("Rahul", "Amit", "Sneha", "Priya")

# Creating a dictionary of students
students_dict = {
    1: "Rahul",
    2: "Amit",
    3: "Sneha",
    4: "Priya"
}

print("List of Students:", students_list)
print("Tuple of Students:", students_tuple)
print("Dictionary of Students:", students_dict)

# ADD operation
students_dict[5] = "Rohan"
print("\nAfter Adding a Student:")
print(students_dict)

# UPDATE operation
students_dict[2] = "Akash"
print("\nAfter Updating a Student:")
print(students_dict)

# DELETE operation
del students_dict[3]
print("\nAfter Deleting a Student:")
print(students_dict)
```
