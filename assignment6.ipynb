```python

import pandas as pd

# Step 1: Read the CSV file
df = pd.read_csv("/content/Titanic-Dataset.csv")

print(df)


```

```python
# Dataset dimensions
print("\nShape:")
print(df.shape)
```

```python
# Column names and data types
print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

```

```python
# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
df.describe()
```

```python
# Step 2: Display the first five rows
print("First Five Rows:")
print(df.head())

# Step 3: Display the last five rows
print("\nLast Five Rows:")
print(df.tail())


```

```python

# Step 7: Display dataset information
print("\nDataset Information:")
df.info()


```

```python

# Step 9: Count missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 10: Count duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

```

```python
# Step 11: Display passenger names
print("\nPassenger Names:")
print(df["Name"].head())

# Step 12: Count passengers by gender
print("\nPassengers by Gender:")
print(df["Sex"].value_counts())

# Step 13: Calculate average age
print("\nAverage Age:")
print(df["Age"].mean())

# Step 14: Find the highest fare
print("\nHighest Fare:")
print(df["Fare"].max())

# Step 15: Find the average fare
print("\nAverage Fare:")
print(df["Fare"].mean())

```

```python
# Step 16: Filter passengers older than 60
print("\nPassengers Older Than 60:")
print(df[df["Age"] > 60][["Name", "Age", "Sex"]])

```

```python
# Step 17: Display passengers who survived
print("\nPassengers Who Survived:")
print(df[df["Survived"] == 1][["Name", "Sex", "Age"]].head())

```

```python
# Step 18: Count survivors and non-survivors
print("\nSurvival Counts:")
print(df["Survived"].value_counts())

```

```python
# Step 19: Calculate survival rate by gender
print("\nSurvival Rate by Gender:")
print(df.groupby("Sex")["Survived"].mean() * 100)

```

```python
# Step 20: Sort passengers by fare
print("\nTop 10 Highest Fares:")
print(
    df.sort_values("Fare", ascending=False)
      [["Name", "Fare"]]
      .head(10)
)

```

```python
# Step 21: Count passengers by passenger class
print("\nPassengers by Class:")
print(df["Pclass"].value_counts().sort_index())



```

```python
# Step 22: Calculate average age by gender
print("\nAverage Age by Gender:")
print(df.groupby("Sex")["Age"].mean())


```

```python
# Step 23: Display embarkation port counts
print("\nPassengers by Embarkation Port:")
print(df["Embarked"].value_counts())
```

```python

# Step 24: Save a copy of the dataset
df.to_csv("titanic_copy.csv", index=False)
```
