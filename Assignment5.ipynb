```python
import re
def check_string(s):
    pattern = "^[A-Za-z0-9]+$" #^ = start,$ = end , += one or more
    return bool(re.match(pattern,s))
s1 = "hello123"
s2 = "hello_123"
print(check_string(s1))
print(check_string(s2))
```

```python
def check_string(s):
    return s.isalnum()
s1="hello123"
s2="hello_123"
print(check_string(s1))
print(check_string(s2))
```
